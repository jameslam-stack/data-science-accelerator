# Aggregating Data With Pandas

---

## Why do we need aggregation?

We aggregate data because **decisions are almost never made at the row level**.  
Instead, we ask higher-level questions such as:

- How much **revenue** did we make per day?
- Which **category** performs best?
- What is the **average order value** by customer or channel?
- How do outcomes differ across **locations** or **time periods**?

---

### What aggregation helps us do

Aggregation allows us to:

- **Reduce noise**  
  Summarise thousands of rows into a few clear, meaningful numbers

- **Compare groups**  
  Spot differences between categories, regions, or time windows

- **Identify patterns and trends**  
  Especially when aggregating over time

- **Prepare data for visualisation and modelling**  
  Most charts and models rely on aggregated inputs

---

```python
# Data question: What does our raw transactional dataset look like before any summarisation?
import pandas as pd

df = pd.DataFrame(
    {
        "order_id":  [1001,1002,1003,1004,1005,1006,1007,1008,1009,1010],
        "date":      pd.to_datetime(["2026-01-02","2026-01-02","2026-01-03","2026-01-03","2026-01-04",
                                     "2026-01-04","2026-01-05","2026-01-05","2026-01-06","2026-01-06"]),
        "customer":  ["C001","C002","C001","C003","C004","C002","C005","C001","C003","C006"],
        "city":      ["London","London","London","Manchester","Manchester","London","Birmingham","London","Manchester","Birmingham"],
        "category":  ["Coffee","Coffee","Tea","Coffee","Tea","Tea","Coffee","Coffee","Tea","Coffee"],
        "qty":       [2,1,3,1,2,1,4,1,2,1],
        "unit_price":[3.50,3.50,2.40,3.50,2.40,2.40,3.50,3.50,2.40,3.50],
    }
)

df["revenue"] = df["qty"] * df["unit_price"]
df
```

# Data question: How many orders do we have in each city?

# Data question: What is the total revenue generated in each city?

```python

df.groupby("city")["revenue"].sum()
```

# Data question: What is the average order revenue in each city?

# Data question: How consistent are order revenues in each city (i.e., how much do they vary)?

```python

df.groupby("city")["revenue"].std()
```

# Introducing `.agg()` — combining multiple aggregates

So far, each aggregation has answered **one question at a time**  
(e.g. `count`, `sum`, `mean`).

In real data analysis, we usually want **several summary metrics together**.

---

### What does `.agg()` do?

The `.agg()` method allows us to:

- Apply **multiple aggregate functions at once**
- Give **clear names** to each aggregated output
- Produce **one tidy summary table** instead of many separate ones



```python
# Data question: For each city, how many orders are there and how much total revenue is generated?

df.groupby("city").agg(
    orders=("order_id", "count"),
    total_revenue=("revenue", "sum"),
)
```

```python
# Data question: How does the typical order compare to the average order in each city?

df.groupby("city").agg(
    avg_order_value=("revenue", "mean"),
    median_order_value=("revenue", "median"),
)
```

```python
# Data question: For each city, how many unique customers do we have and
# what is the highest single-order revenue recorded?

df.groupby("city").agg(
    customers=("customer", "nunique"),
    max_order_value=("revenue", "max"),
)
```

In standard Python, we use `for` loops. In Data Science, looping over 10 million rows is too slow. Here we introduce numpy and show the improvements from  **Vectorization**."

### 1 Vectorized Math vs. Loops
**Scenario:** Calculate the Natural Log `ln()` of revenue.

##### Generate big data for the test (1 Million Rows)

```python
import numpy as np
import pandas as pd
import time
import math
import numexpr

# Generate big data for the test (10 Million Rows)
big_data = np.random.rand(100000000) * 1000
df_big = pd.DataFrame({'revenue': big_data})
df_big.head(5)
```

```python
big_data.shape
```

#### Method 1: The Loop

```python
start = time.time()
res = []
for x in df_big['revenue']:
    res.append(math.log(x))
end = time.time()
print(f"Loop Time: {end - start:.4f} seconds")
```

```python
# --- APPROACH 2: Numpy Vectorization (Fast) ---
start = time.time()
df_big['log_revenue'] = np.log(df_big['revenue'])
end = time.time()
print(f"Numpy Time: {end - start:.4f} seconds")

# Instructor Note: Point out that Numpy is usually 50-100x faster.
```

# FUNCTIONS FOR DATA PREPROCESSING

---

## Why do we need functions?

Before introducing functions, it’s important to understand **the problem they are solving**.

In real data work — especially during **data preprocessing** — you will often:

- Clean **multiple columns in the same way**
- Copy and paste **working code** to move faster
- Make **small tweaks** as requirements evolve

At first, this approach feels efficient.

---

### The hidden problem

As your dataset grows and your logic changes, this pattern quietly introduces risks:

- Repeated logic becomes **hard to update**
- Small changes can lead to **inconsistent behaviour**
- Bugs are more likely to appear in **only some places**

---

### Why this example is written this way

This example is written for you as a **data analyst transitioning toward data science**.

It deliberately **starts by repeating code**, so you can:

- Clearly see the **pain points**
- Feel why the approach stops scaling
- Understand **why functions exist**, before learning how to write them




# EXAMPLE

```python
import pandas as pd
import numpy as np

# Simulated raw data
df = pd.DataFrame({
    "city_a": [" London ", "PARIS", None, "  berlin"],
    "city_b": [" madrid", None, "ROME ", " Lisbon"],
    "city_c": [None, " ATHENS", "vienna ", "  Prague"]
})

df
```

Repeating the Same Cleaning Logic (❌ Not Scalable)


```python
# Clean city_a
df["city_a"] = (
    df["city_a"]
    .fillna("unknown")
    .str.strip()
    .str.lower()
)

# Clean city_b
df["city_b"] = (
    df["city_b"]
    .fillna("unknown")
    .str.strip()
    .str.lower()
)

# Clean city_c
df["city_c"] = (
    df["city_c"]
    .fillna("unknown")
    .str.strip()
    .str.lower()
)

df
```


# Problems with this approach



- Code duplication
- Hard to maintain (change logic in 3 places)
- Easy to introduce inconsistencies

# Functional approach


```python
def clean_text_column(series: pd.Series) -> pd.Series:
    return (
        series
        .fillna("unknown")
        .str.strip()
        .str.lower()
    )
```

```python
for col in ["city_a", "city_b", "city_c"]:
    df[col] = clean_text_column(df[col])

df
```

# Why does this matter?

- Single source of truth for preprocessing logic

- Easier experimentation (change once, apply everywhere)

- Functions compose cleanly into pipelines later (e.g. sklearn)

- Clear boundary between what you do and where you do it

- This is the first step toward reproducible, testable preprocessing pipelines.

# Function Syntax

```python
def function_name(parameter1, parameter2):
    """Optional description of what the function does."""
    # function body (indented)
    result = parameter1 + parameter2
    return result
```

## Python Function Anatomy

**def**  
- Short for *define*  
- Tells Python you are creating a function  

**function_name**  
- How you refer to and call the function later  
- Should describe what the function does  

**(parameter1, parameter2)**  
- Parameters are inputs to the function  
- Both inputs must be provided when calling the function  

**:**  
- Marks the start of the function body  

**Indentation**  
- Everything indented underneath belongs to the function  
- This is how Python knows what code is inside the function  

**return**  
- Sends a value back to where the function was called  
- Ends the function immediately

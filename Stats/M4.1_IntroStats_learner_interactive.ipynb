## ✅ Key takeaways

* Probability is estimated by counting rows.
* Hypothesis tests check evidence against a null assumption.
* Welch's t-test and chi-square test are common Python tools.
* Confidence intervals give a range of plausible values.
* Effect size matters more than p-value for decisions.
* Correlation is not causation - watch for confounders.
* Simpson's paradox: always check subgroups.
* CLT justifies normal-based methods for large samples.
* Bootstrap works when distributions are unknown.
* Standardization, CV, and leakage awareness are essential for ML.

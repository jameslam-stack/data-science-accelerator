# Section 1 Train Test Split

scenario:


- An analyst creates several new features (ratios, logs, aggregates) using the full dataset and then performs a train/test split.

Question:

- What is the hidden risk in this workflow, and why does it count as data leakage?

A. The model will overfit because feature engineering increases the number of features.

B. Information from the test set influences how features are created, meaning the training data is no longer independent of the test data.

C. Feature engineering should only be done after model training to avoid bias.

✅ Correct answer: B

Why:

Feature engineering learns from data. If it’s fit on the full dataset, the test set has already influenced the training pipeline.







Scenario:

- A student computes “average spend per customer” using the full dataset and merges it back before splitting.

Question:

- Why is this especially dangerous for supervised learning?

A. Aggregates reduce variance, which can hide useful information.

B. The aggregate feature indirectly encodes information about the target across both training and test data.

C. Group-based features only work for unsupervised learning tasks.

✅ Correct answer: B

Why:

Group aggregates are highly predictive. Computing them on the full dataset leaks future (test) behaviour into training, inflating performance.






Scenario:

- A dataset with a rare target class is split randomly without stratification.

Question:

- What problem might appear in the test set, and how would this affect evaluation?

A. The test set may contain very few or no examples of the rare class, making performance metrics unreliable.

B. The model will fail to train because rare classes require special algorithms.

C. Random splits always improve generalisation by increasing randomness.

✅ Correct answer: A

Why:
Without stratification, the test set may not represent the real class distribution, leading to misleading accuracy and unstable metrics.


# Section 2 EDA and Encoding

Scenario:

- After splitting the data, an analyst explores distributions and correlations in the test set to decide whether to apply transformations.

Question:
- Why is this considered a workflow mistake?

A. The test set should only be used for final evaluation, not to inform preprocessing decisions.
B. EDA should only be done after a model has been trained.
C. The test set is usually too small to visualise properly.

✅ Correct answer: A

Why (teacher reference):
Even without fitting a model, decisions influenced by the test set break the idea of “unseen data.”



Scenario:
- An analyst removes one of two features simply because they are highly correlated, without considering the model.

Question:
- What is the main issue with this approach?

A. Correlated features always cause models to fail.
B. Correlation alone doesn’t tell you whether a feature is useful for the chosen model.
C. Correlation should only be measured after scaling.

✅ Correct answer: B

Why (teacher reference):
Correlation ≠ useless. Whether correlated features are a problem depends on the model and goal.



Scenario:

A student applies label encoding to a “city” column before training a linear model.

Question:
What unintended assumption does this introduce?

A. The model assumes the cities have a meaningful numeric order or ranking.
B. The model assumes all cities appear with equal frequency.
C. The model assumes the city column should be treated as continuous data.

✅ Correct answer: A

Teacher note:
Label encoding turns categories into numbers, and linear models treat numbers as ordered.
This implies relationships like City A < City B < City C, which do not exist for nominal data.
Use one-hot encoding for nominal categories.

# Section 3 Transformation And Scaling



Scenario:
An analyst standardizes all numeric columns, including binary flags (0/1).

Question:
Why might this be unnecessary or harmful?

A. Scaling binary features removes their clear meaning and can make them harder for the model to interpret.
B. Binary features cannot be used by machine learning models once scaled.
C. Standardization always reduces model accuracy when binary data is present.

✅ Correct answer: A

Teacher note:
Binary features already live on a meaningful scale. Standardizing them adds noise without adding information.


Scenario:
A log transform is applied automatically to all positive numeric features.

Question:
What assumption is being made, and when might this backfire?

A. It assumes all positive features are right-skewed; this can distort features that are already well-behaved.
B. It assumes log transforms always improve linear models.
C. It assumes log transforms reduce missing values in the data.

✅ Correct answer: A

Teacher note:
Log transforms are useful for skewed data, not a default step. Applying them blindly can harm signal and interpretability.

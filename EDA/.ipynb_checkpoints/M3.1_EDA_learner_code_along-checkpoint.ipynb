````python
# Task 11: Complete the method chain.
# Hints:
# - Use copy to avoid changing the original data.
# - Use assign to create or update columns.
# - Use fillna to fill missing values.

analysis_df = (
    performance_df
    .___()
    .assign(
        programme=lambda df: df["programme"].str.strip().str.title(),
        region=lambda df: df["region"].str.strip().str.upper(),
        attendance_rate=lambda df: df["attendance_rate"].___(df["attendance_rate"].median()),
        engagement_score=lambda df: df["engagement_score"].___(df["engagement_score"].median()),
        submission_rate=lambda df: df["submission_rate"].___(df["submission_rate"].median())
    )
    .assign(
        engagement_gap=lambda df: df["engagement_score"] - 70
    )
)

analysis_df.head()
````

## 19. Correlation

**Goal:** Explore whether numeric columns move together.

Correlation values range from `-1` to `1`.

- Close to `1`: values tend to increase together.
- Close to `-1`: one value tends to decrease as the other increases.
- Close to `0`: no clear straight-line relationship.

Correlation does not prove that one thing causes another.

````python
# Task 19: Create a correlation table.
# Complete the method at the end.

numeric_for_correlation = [
    "attendance_rate",
    "engagement_score",
    "submission_rate",
    "support_sessions",
    "assessment_score",
    "satisfaction_score",
    "engagement_gap"
]

correlation_table = enriched_df[numeric_for_correlation].___().round(2)
correlation_table
````

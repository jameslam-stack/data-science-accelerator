## 19. Higher-order functions


### Decorator function
Create function that wraps another function


```python
def doubler(f):
    def g(x):
        return 2 * f(x)
    return g
```

### Simple function
Define function to be decorated

```python
def f1(x):
    return x + 1
```

### Test decorated function
Verify wrapper function works correctly

```python
g = doubler(f1)
```

```python
assert g(3) == 8
```

```python
assert g(-1) == 0
```

### Two-parameter function
Function with multiple parameters

```python
def f2(x, y):
    return x + y
```

### Apply decorator
Wrap two-parameter function


```python
g = doubler(f2)
```

### Error with wrong arguments
Wrapper only accepts one argument - error handling

```python
try:
    g(1, 2)
except TypeError:
    print("as defined, g only takes one argument")
```

### Corrected decorator
Use `*args, **kwargs` to handle any arguments


```python
def doubler_correct(f):
    def g(*args, **kwargs):
        return 2 * f(*args, **kwargs)
    return g
```


### Test corrected version
Verify wrapper now works with multiple parameters

```python
g = doubler_correct(f2)
assert g(1, 2) == 6
```

## Class definition
Define class with methods including `__init__` and `__repr__`

```python
class CountingClicker:
    def __init__(self, count=0):
        self.count = count

    def __repr__(self):
        return f"CountingClicker(count={self.count})"

    def click(self, num_times=1):
        self.count += num_times

    def read(self):
        return self.count

    def reset(self):
        self.count = 0
```

### Test methods
Call class methods and verify behavior

```python
assert clicker.read() == 0
clicker.click()
clicker.click()
assert clicker.read() == 2
clicker.reset()
assert clicker.read() == 0
```

### Inheritance
Define subclass that overrides parent method

```python
class NoResetClicker( CountingClicker):
    def reset(self):
        pass
```

### Test subclass

## 21. Generators

### Generator function
Define function using `yield` for lazy evaluation

```python
def generate_range(n):
    i = 0
    while i < n:
        yield i
        i += 1
```

### Iterate generator
Loop through yielded values


```python
for i in generate_range(10):
    print(f"i: {i}")
```


### Infinite generator
Create generator with infinite sequence

```python
def natural_numbers():
    n = 1
    while True:
        yield n
        n += 1

```

### Generator expression
Lazy evaluation using comprehension syntax


```python
evens_below_20 = (i for i in generate_range(20) if i % 2 == 0)
```

### Chained generators
Combine multiple generator expressions for efficient processing

```python
data = natural_numbers()
evens = (x for x in data if x % 2 == 0)
even_squares = (x ** 2 for x in evens)
even_squares_ending_in_six = (x for x in even_squares if x % 10 == 6)
```

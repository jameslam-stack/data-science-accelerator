## Activity 1: Data detective - data cleaning and imputation

### Tasks:
1. Load `uk_air_quality.csv` into Pandas.
2. Convert the date to a datetime object using Pandas library.
3. Set the index of the data frame and sort it chronologically.
4. Use `.asfreq('D')` to expose missing calendar days.
5. Replace invalid readings (values `< 0` and values `> 200`) with `np.nan`.
6. Fill missing values using a suitable method

```python
import numpy as np
import pandas as pd

# 1. Load data
df = pd.read_csv("uk_air_quality.csv")
```

```python

# 2 & 3. Convert to datetime, set as index, and sort
df["reading_date"] = pd.to_datetime(
    df["reading_date"], format="mixed", dayfirst=True
)
df = df.set_index("reading_date").sort_index()
```

```python

# 4. Reveal missing dates
df_grid = df.asfreq("D")
```

```python
# 5. Filter out invalid values
df_grid.loc[df_grid["pm25_concentration"] < 0, "pm25_concentration"] = np.nan
df_grid.loc[df_grid["pm25_concentration"] > 200, "pm25_concentration"] = np.nan

```

```python
# 6. Fill missing values
df_cleaned = df_grid.copy()
df_cleaned["pm25_concentration"] = df_cleaned["pm25_concentration"].interpolate()

```

```python
# View cleaned dataset
df_cleaned.head(10)
```

## Activity 2: Granularity vs. Noise - Resampling your data

### Tasks:
1. Select the cleaned `pm25_concentration` series from Activity 1.
2. Calculate a **7-day rolling mean** .
3. Calculate a **monthly mean** `resample()`.
4. Plot raw daily data, the 7-day rolling mean, and the monthly mean on a single line chart.

```python
import matplotlib.pyplot as plt

# 1. Select the cleaned series
pm25_series = df_cleaned["pm25_concentration"]
```

```python

# 2. Calculate 7-day rolling mean
rolling_7d = pm25_series.rolling(window=7).mean()
```

```python

# 3. Calculate monthly mean
monthly_mean = pm25_series.resample("ME").mean()
```

```python

# 4. Plot results
plt.figure(figsize=(10, 5))
pm25_series.plot(label="Raw Daily Data", alpha=0.4, color="gray")
rolling_7d.plot(label="7-Day Rolling Mean", color="blue")
monthly_mean.plot(label="Monthly Resampled Mean", color="red", marker="o")

plt.title("UK Air Quality: Daily vs. Rolling vs. Resampled")
plt.ylabel("PM2.5 Concentration")
plt.legend()
plt.grid(True)
plt.show()
```
